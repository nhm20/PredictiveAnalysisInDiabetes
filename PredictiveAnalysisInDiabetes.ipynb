```python
# Insulin - Right skewed, so we will use Median for imputation
if not dataframe['Insulin'].replace(0, np.nan).dropna().empty:
    dataframe['Insulin'] = dataframe['Insulin'].replace(0, dataframe['Insulin'].median())

if not dataframe['Pregnancies'].replace(0, np.nan).dropna().empty:
    dataframe['Pregnancies'] = dataframe['Pregnancies'].replace(0, dataframe['Pregnancies'].median())

if not dataframe['Glucose'].replace(0, np.nan).dropna().empty:
    dataframe['Glucose'] = dataframe['Glucose'].replace(0, dataframe['Glucose'].mean())

if not dataframe['BloodPressure'].replace(0, np.nan).dropna().empty:
    dataframe['BloodPressure'] = dataframe['BloodPressure'].replace(0, dataframe['BloodPressure'].mean())

if not dataframe['SkinThickness'].replace(0, np.nan).dropna().empty:
    dataframe['SkinThickness'] = dataframe['SkinThickness'].replace(0, dataframe['SkinThickness'].median())

if not dataframe['BMI'].replace(0, np.nan).dropna().empty:
    dataframe['BMI'] = dataframe['BMI'].replace(0, dataframe['BMI'].mean())

if not dataframe['DiabetesPedigreeFunction'].replace(0, np.nan).dropna().empty:
    dataframe['DiabetesPedigreeFunction'] = dataframe['DiabetesPedigreeFunction'].replace(0, dataframe['DiabetesPedigreeFunction'].median())

if not dataframe['Age'].replace(0, np.nan).dropna().empty:
    dataframe['Age'] = dataframe['Age'].replace(0, dataframe['Age'].median())
```

```python
##X -> input features and y -> target variable
X = dataframe.drop(columns='Outcome', axis=1)
y= dataframe['Outcome']
```

Outlier Detection -> Box Plot

```python
import matplotlib.pyplot as plt
fig, ax = plt.subplots(figsize = (15,15))
sns.boxplot(data=X, ax=ax)
plt.savefig('boxplot.png')
plt.show()
```

```python
X.shape
```

```python
y.shape
```

```python
cols= ['Pregnancies','Glucose','BloodPressure','SkinThickness','Insulin','BMI','DiabetesPedigreeFunction','Age']
for col in cols:
    Q1= X[col].quantile(0.25)
    Q3=X[col].quantile(0.75)
    IQR=Q3-Q1
    lower_whisker=Q1-1.5*IQR
    upper_whisker=Q3+1.5*IQR
    mask= (X[col]>=lower_whisker) & (X[col]<=upper_whisker)
```

```python
X_outliers_removed = X[mask]
y_outliers_removed = y[mask]
```
