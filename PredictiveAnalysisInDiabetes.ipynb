```python
# Insulin - Right skewed, so we will use Median for imputation
if not dataframe['Insulin'].replace(0, np.nan).dropna().empty:
    dataframe['Insulin'] = dataframe['Insulin'].replace(0, dataframe['Insulin'].median())

if not dataframe['Pregnancies'].replace(0, np.nan).dropna().empty:
    dataframe['Pregnancies'] = dataframe['Pregnancies'].replace(0, dataframe['Pregnancies'].median())

if not dataframe['Glucose'].replace(0, np.nan).dropna().empty:
    dataframe['Glucose'] = dataframe['Glucose'].replace(0, dataframe['Glucose'].mean())

if not dataframe['BloodPressure'].replace(0, np.nan).dropna().empty:
    dataframe['BloodPressure'] = dataframe['BloodPressure'].replace(0, dataframe['BloodPressure'].mean())

if not dataframe['SkinThickness'].replace(0, np.nan).dropna().empty:
    dataframe['SkinThickness'] = dataframe['SkinThickness'].replace(0, dataframe['SkinThickness'].median())

if not dataframe['BMI'].replace(0, np.nan).dropna().empty:
    dataframe['BMI'] = dataframe['BMI'].replace(0, dataframe['BMI'].mean())

if not dataframe['DiabetesPedigreeFunction'].replace(0, np.nan).dropna().empty:
    dataframe['DiabetesPedigreeFunction'] = dataframe['DiabetesPedigreeFunction'].replace(0, dataframe['DiabetesPedigreeFunction'].median())

if not dataframe['Age'].replace(0, np.nan).dropna().empty:
    dataframe['Age'] = dataframe['Age'].replace(0, dataframe['Age'].median())
```

```python
##X -> input features and y -> target variable
X = dataframe.drop(columns='Outcome', axis=1)
y= dataframe['Outcome']
```

Outlier Detection -> Box Plot

```python
import matplotlib.pyplot as plt
fig, ax = plt.subplots(figsize = (15,15))
sns.boxplot(data=X, ax=ax)
plt.savefig('boxplot.png')
plt.show()
```

```python
X.shape
```

```python
y.shape
```

```python
cols= ['Pregnancies','Glucose','BloodPressure','SkinThickness','Insulin','BMI','DiabetesPedigreeFunction','Age']
for col in cols:
    Q1= X[col].quantile(0.25)
    Q3=X[col].quantile(0.75)
    IQR=Q3-Q1
    lower_whisker=Q1-1.5*IQR
    upper_whisker=Q3+1.5*IQR
    mask= (X[col]>=lower_whisker) & (X[col]<=upper_whisker)
```

```python
X_outliers_removed = X[mask]
y_outliers_removed = y[mask]
```

```python
X_outliers_removed.shape
```


Standardization of Data

Approach 2: Quantiles

```python
X_scaled.reset_index(drop=True, inplace=True)
y_outliers_removed.reset_index(drop=True, inplace=True)
```

```python
q=X_scaled['Insulin'].quantile(0.95)
mask= X_scaled['Insulin']<q
dataNew= X_scaled[mask]
y_outliers_removed= y_outliers_removed[mask]
dataNew.shape
```

Model Training

Splitting the data into training and testing sets

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(dataNew, y_outliers_removed, test_size=0.33, random_state=42)
```

```python
X_train.shape
```

```python
X_test.shape
```

Data imbalancing - techniques
- oversampling : Minority class and  increased that number to majority class
- undersampling : Majority class and  decreased that number to minority class
- SMOTE (Synthetic Minority Over-sampling Technique) : Synthetic data and increate the no of samples to majority class

```python
y_train.value_counts()
```

SMOTE Technique

```python
from imblearn.over_sampling import SMOTE
smote = SMOTE(random_state=42)
X_train_resampled, y_train_resampled = smote.fit_resample(X_train, y_train)
print(pd.Series(y_train_resampled).value_counts())
```

Model Training

```python
from sklearn.linear_model import LogisticRegression
classification=LogisticRegression()
classification.fit(X_train_resampled, y_train_resampled)
```

Model Prediction

```python
y_predictions=classification.predict(X_test)
print(y_predictions)
```

Model Evaluation

```python
from sklearn.metrics import accuracy_score
accuracy_score(y_test,y_predictions)
```

Healthcare: Recall is more important than precision

```python
from sklearn.metrics import classification_report
target_names = ['Non-Diabetic', 'Diabetic']
print(classification_report(y_test, y_predictions, target_names=target_names))
```
